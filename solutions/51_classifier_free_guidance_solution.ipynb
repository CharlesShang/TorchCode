```python
import torch
```

```python
# ✅ SOLUTION

def classifier_free_guidance(uncond: torch.Tensor, cond: torch.Tensor,
                             guidance_scale: float = 7.5, rescale: float = 0.0) -> torch.Tensor:
    guided = uncond + guidance_scale * (cond - uncond)
    if rescale > 0.0:
        dims = tuple(range(1, guided.ndim))
        guided_std = guided.std(dim=dims, keepdim=True).clamp_min(1e-6)
        cond_std = cond.std(dim=dims, keepdim=True)
        rescaled = guided * (cond_std / guided_std)
        guided = rescale * rescaled + (1.0 - rescale) * guided
    return guided
```

```python
# Verify
uncond = torch.tensor([0.0, 1.0])
cond = torch.tensor([2.0, 3.0])
print(classifier_free_guidance(uncond, cond, guidance_scale=2.0))
```
