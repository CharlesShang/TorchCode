```python
import torch
```

```python
# ✅ SOLUTION

def sample_logit_normal_timesteps(batch_size: int, mean: float = 0.0, std: float = 1.0,
                                  device=None, dtype=torch.float32, generator=None) -> torch.Tensor:
    z = torch.randn(batch_size, device=device, dtype=dtype, generator=generator) * std + mean
    return torch.sigmoid(z)
```

```python
# Verify
g = torch.Generator().manual_seed(0)
t = sample_logit_normal_timesteps(8, generator=g)
print(t)
print(t.min().item(), t.max().item())
```
