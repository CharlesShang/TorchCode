```python
import torch
```

```python
# ✅ SOLUTION

def rgb_to_grayscale(x: torch.Tensor, keepdim: bool = True,
                     weights: torch.Tensor | None = None) -> torch.Tensor:
    if x.shape[-3] != 3:
        raise ValueError("expected RGB channel dimension of size 3")
    if weights is None:
        weights = torch.tensor([0.299, 0.587, 0.114], device=x.device, dtype=x.dtype)
    else:
        weights = weights.to(device=x.device, dtype=x.dtype)
    view_shape = (1,) * (x.ndim - 3) + (3, 1, 1)
    gray = (x * weights.view(view_shape)).sum(dim=-3, keepdim=keepdim)
    return gray
```

```python
# Verify
x = torch.rand(2, 3, 4, 4)
gray = rgb_to_grayscale(x)
print(gray.shape)
```
