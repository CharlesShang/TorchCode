```python
import torch
```

```python
# ✅ SOLUTION

def uniform_temporal_sample(video: torch.Tensor, num_frames: int,
                            temporal_dim: int = 0) -> torch.Tensor:
    T = video.shape[temporal_dim]
    if num_frames <= 0:
        raise ValueError("num_frames must be positive")
    idx = torch.linspace(0, T - 1, num_frames, device=video.device).round().long()
    return video.index_select(temporal_dim, idx)
```

```python
# Verify
video = torch.arange(10 * 1 * 1 * 1).view(10, 1, 1, 1)
print(uniform_temporal_sample(video, 4).flatten())
```
