```python
import torch
```

```python
# ✅ SOLUTION

class FP8Quantizer:
    def __init__(self, max_fp8: float = 448.0):
        self.max_fp8 = float(max_fp8)

    def quantize(self, x: torch.Tensor):
        amax = x.abs().max()
        scale = (amax / self.max_fp8).clamp_min(1e-12)
        q = torch.round(x / scale).clamp(-self.max_fp8, self.max_fp8).to(torch.int16)
        return q, scale

    def dequantize(self, q: torch.Tensor, scale: torch.Tensor):
        return q.to(scale.dtype) * scale
```

```python
# Verify
q = FP8Quantizer()
x = torch.randn(8) * 3
codes, scale = q.quantize(x)
print(codes, scale)
print(q.dequantize(codes, scale))
```
