```python
import threading
```

```python
# ✅ SOLUTION

class ThreadSafeCounter:
    def __init__(self, initial: int = 0):
        self._value = initial
        self._lock = threading.Lock()

    def increment(self, n: int = 1) -> int:
        with self._lock:
            self._value += n
            return self._value

    def value(self) -> int:
        with self._lock:
            return self._value
```

```python
# Verify
counter = ThreadSafeCounter()
print(counter.increment())
print(counter.value())
```
