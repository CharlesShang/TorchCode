```python
import torch
```

```python
# ✅ SOLUTION

def euler_flow_sample(model, x0: torch.Tensor, steps: int = 20, return_trajectory: bool = False):
    if steps <= 0:
        raise ValueError("steps must be positive")
    x = x0.clone()
    dt = 1.0 / steps
    trajectory = [x.clone()] if return_trajectory else None
    B = x.shape[0]
    for i in range(steps):
        t = torch.full((B,), i / steps, device=x.device, dtype=x.dtype)
        velocity = model(x, t)
        x = x + dt * velocity
        if return_trajectory:
            trajectory.append(x.clone())
    return trajectory if return_trajectory else x
```

```python
# Verify
class ConstantVelocity:
    def __call__(self, x, t):
        return torch.ones_like(x) * 2

x0 = torch.zeros(2, 3)
print(euler_flow_sample(ConstantVelocity(), x0, steps=4))
```
