# 🟡 Solution: Sinusoidal Positional Encoding

Reference solution for deterministic Transformer positional encodings.

```python
import torch
import math
```

```python
# ✅ SOLUTION

def sinusoidal_positional_encoding(seq_len: int, dim: int, base: float = 10000.0,
                                   device=None, dtype=torch.float32) -> torch.Tensor:
    position = torch.arange(seq_len, device=device, dtype=dtype).unsqueeze(1)
    div_term = torch.exp(
        torch.arange(0, dim, 2, device=device, dtype=dtype) *
        (-math.log(base) / dim)
    )

    pe = torch.zeros(seq_len, dim, device=device, dtype=dtype)
    angles = position * div_term
    pe[:, 0::2] = torch.sin(angles)
    pe[:, 1::2] = torch.cos(angles[:, :pe[:, 1::2].shape[1]])
    return pe
```

```python
# Verify
pe = sinusoidal_positional_encoding(seq_len=4, dim=6)
print(pe)
print("Shape:", pe.shape)
print("First row:", pe[0])  # even columns 0, odd columns 1
```
