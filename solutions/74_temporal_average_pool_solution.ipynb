# 🟡 Solution: Temporal Average Pooling for Video

```python
import torch
import torch.nn.functional as F
```

```python
# ✅ SOLUTION

def temporal_avg_pool(video: torch.Tensor, kernel_size: int,
                      stride: int | None = None, padding: int = 0) -> torch.Tensor:
    if stride is None:
        stride = kernel_size
    if padding:
        video = F.pad(video, (0, 0, 0, 0, padding, padding))
    windows = video.unfold(2, kernel_size, stride)
    return windows.mean(dim=-1)
```

```python
# Verify
video = torch.randn(2, 3, 8, 4, 4)
print(temporal_avg_pool(video, 3, stride=2, padding=1).shape)
```
