# 🟡 Medium: Patchify / Unpatchify Latents

Implement latent **patchify / unpatchify** utilities used by ViT, DiT, and MMDiT backbones.

Given an image latent tensor `x` with shape `(B, C, H, W)`, split it into non-overlapping patches and flatten every patch into a token. Then reconstruct the original latent from those tokens.

### Signature
```python
class PatchifyLatents:
    def __init__(self, patch_size): ...
    def patchify(self, x: torch.Tensor) -> torch.Tensor: ...
    def unpatchify(self, tokens: torch.Tensor, image_size, channels: int) -> torch.Tensor: ...
```

### Requirements
- `patch_size` can be an `int` or `(patch_h, patch_w)` tuple
- `patchify(x)` returns shape `(B, num_patches, C * patch_h * patch_w)`
- `unpatchify(tokens, image_size, channels)` returns shape `(B, C, H, W)`
- Use basic tensor reshapes/permutations; do not use convolution
