# 🔴 Hard: Paged KV Cache

Implement a small **paged KV cache** simulator inspired by vLLM.

### Signature
```python
class PagedKVCache:
    def __init__(self, num_layers, num_heads, head_dim, block_size, max_blocks,
                 device=None, dtype=torch.float32): ...
    def append(self, layer_id: int, seq_id: str, key: Tensor, value: Tensor): ...
    def get(self, layer_id: int, seq_id: str) -> tuple[Tensor, Tensor]: ...
    def free(self, seq_id: str): ...
```

### Requirements
- Store K/V in fixed-size blocks
- `key` and `value` chunks have shape `(T, num_heads, head_dim)`
- Maintain a page table per sequence
- Support appending across block boundaries
- `get` should return exactly the appended tokens, in order
- `free` should release the sequence's blocks for reuse
