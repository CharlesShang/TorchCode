# 🟡 Medium: Prefix Cache

Implement a simple **prefix cache** for prompt KV reuse.

### Signature
```python
class PrefixCache:
    def add(self, prompt_tokens, key: Tensor, value: Tensor): ...
    def lookup(self, prompt_tokens) -> tuple[int, Tensor | None, Tensor | None]: ...
```

### Requirements
- Cache keys by token prefix
- `lookup` returns the longest cached prefix that matches the requested prompt
- Return `(matched_len, key_prefix, value_prefix)`
- If no prefix matches, return `(0, None, None)`
- Return cloned tensors so callers cannot mutate the cache accidentally
