# 🔴 Hard: Speculative Decoding with Verification

Implement greedy **speculative decoding with target verification**.

This is an interview-sized deterministic version: the draft model proposes tokens greedily, and the target model verifies greedily.

### Signature
```python
def speculative_decode_verify(target_model, draft_model, prefix: list[int], max_new_tokens: int,
                              draft_steps: int = 4, eos_token_id: int | None = None) -> list[int]:
    ...
```

### Requirements
- Models are callables: `model(tokens) -> logits_for_next_token`
- Draft proposes up to `draft_steps` tokens with argmax
- Target verifies each proposed token with argmax from the current accepted prefix
- Accept matching draft tokens
- On first mismatch, append the target token and restart drafting
- Stop after `max_new_tokens` new tokens or EOS
