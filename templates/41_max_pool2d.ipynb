# 🟡 Medium: MaxPool2d

Implement **2D max pooling** from scratch for image tensors.

Given `x` with shape `(N, C, H, W)`, slide a window over the spatial dimensions and return the maximum value from each window independently for every batch item and channel.

### Signature
```python
def my_max_pool2d(
    x: torch.Tensor,
    kernel_size,
    stride=None,
    padding=0,
) -> torch.Tensor:
    ...
```

### Requirements
- Support `kernel_size`, `stride`, and `padding` as either an `int` or a `(height, width)` tuple
- If `stride is None`, use `stride = kernel_size`
- Padding should behave like PyTorch max pooling: padded values should be `-inf`, not zero
- Do **NOT** use `torch.nn.MaxPool2d` or `torch.nn.functional.max_pool2d`

### Example
```python
x = torch.arange(1., 17.).view(1, 1, 4, 4)
my_max_pool2d(x, kernel_size=2)
# tensor([[[[ 6.,  8.],
#           [14., 16.]]]])
```
