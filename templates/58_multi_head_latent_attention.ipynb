# 🔴 Hard: Multi-Head Latent Attention

Implement a compact educational version of **Multi-Head Latent Attention (MLA)**.

MLA stores a compressed latent KV representation, then up-projects it to keys and values during attention.

### Signature
```python
class MultiHeadLatentAttention(nn.Module):
    def __init__(self, dim: int, num_heads: int, latent_dim: int): ...
    def forward(self, x: Tensor, latent_cache: Tensor | None = None) -> tuple[Tensor, Tensor]: ...
```

### Requirements
- `x`: `(B, S, D)`
- Compress current tokens with `kv_down: D -> latent_dim`
- If `latent_cache` is provided, append current latent tokens along sequence dim
- Up-project all latent tokens into K and V
- Return `(output, new_latent_cache)`
