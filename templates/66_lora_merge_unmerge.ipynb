# 🟡 Medium: LoRA Merge / Unmerge

Implement a **mergeable LoRA linear layer**.

```python
class MergeableLoRALinear(nn.Module):

    def forward(self, x): ...
    def merge(self): ...
    def unmerge(self): ...
```

### Requirements
- Base `weight` is frozen (`requires_grad=False`)
- LoRA matrices `lora_A` and `lora_B` are trainable
- Effective update is `(alpha / r) * (lora_B @ lora_A)`
- `merge()` adds the update to base weight and sets `merged=True`
- `unmerge()` reverses that exactly
