# 🔴 Hard: Continuous Batching Scheduler

Implement a tiny **continuous batching scheduler** for LLM serving.

### Signature
```python
class ContinuousBatchingScheduler:
    def __init__(self, max_batch_size: int): ...
    def add_request(self, request_id: str, prompt_tokens: list[int], max_new_tokens: int): ...
    def step(self) -> dict: ...
    def has_pending(self) -> bool: ...
```

### Step behavior
Each call returns:
```python
{"prefill": [...], "decode": [...], "finished": [...]}
```

### Requirements
- New requests must run exactly one prefill step before decoding
- Each decode step generates one token for that request
- Requests finish after `max_new_tokens` decode steps
- Schedule at most `max_batch_size` total requests per step
- Prefer prefill requests before decode requests so new traffic can join continuously
