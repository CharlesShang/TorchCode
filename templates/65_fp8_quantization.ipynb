# 🟡 Medium: FP8 Quantization Basics

Implement an educational **FP8-style dynamic quantizer**.

This task focuses on the scaling logic used around FP8 kernels rather than hardware FP8 dtypes.

### Signature
```python
class FP8Quantizer:
    def __init__(self, max_fp8: float = 448.0): ...
    def quantize(self, x: Tensor) -> tuple[Tensor, Tensor]: ...
    def dequantize(self, q: Tensor, scale: Tensor) -> Tensor: ...
```

### Requirements
- Dynamic scale: `scale = max(abs(x)) / max_fp8`
- Quantized codes are rounded and clamped to `[-max_fp8, max_fp8]`
- Return codes as integer tensor and scale as scalar tensor
- Dequantize with `q.float() * scale`
