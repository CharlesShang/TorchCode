# 🔴 Hard: Video Tubelet Patchify

Implement video tubelet patchification for video transformers.

### Signature
```python
class VideoTubeletPatcher:
    def __init__(self, tubelet_size: int, patch_size: int): ...
    def patchify(self, video: torch.Tensor) -> torch.Tensor: ...
    def unpatchify(self, tokens: torch.Tensor, video_shape: tuple) -> torch.Tensor: ...
```

### Requirements
- Input video shape is `(B, T, C, H, W)`
- Output tokens shape is `(B, N, tubelet_size * C * patch_size * patch_size)`
- `T`, `H`, and `W` must be divisible by tubelet/patch sizes
- `unpatchify` must exactly reconstruct the video
