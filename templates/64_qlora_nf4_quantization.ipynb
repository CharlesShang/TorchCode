Implement educational **NF4 block quantization**, the core quantization idea behind QLoRA.

### Signature
```python
class NF4Quantizer:
    def __init__(self, block_size: int = 64): ...
    def quantize(self, weight: Tensor) -> tuple[Tensor, Tensor, tuple]: ...
    def dequantize(self, codes: Tensor, scales: Tensor, original_shape: tuple) -> Tensor: ...
```

### Requirements
- Use a fixed 16-value NF4-like codebook
- Quantize per block with `scale = max(abs(block))`
- Return `codes` as `torch.uint8`, values in `[0, 15]`
- Return one scale per block
- Dequantization should restore the original shape
