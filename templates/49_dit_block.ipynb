### Requirements
- `x`: `(B, S, D)`, `cond`: `(B, cond_dim)`
- Use self-attention with `num_heads`
- Use MLP hidden size `mlp_ratio * dim` and GELU
- Use AdaLN-Zero style `shift/scale/gate` for attention and MLP branches
- Zero-initialize only the modulation linear layer so the whole block starts as identity
