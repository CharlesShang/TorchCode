# 🟡 Medium: AdaLN-Zero Modulation

Implement **AdaLN-Zero** modulation, the conditioning mechanism used in DiT-style diffusion transformer blocks.

The module should normalize tokens, then produce two sets of adaptive `(shift, scale, gate)` values from a conditioning vector.

### Signature
```python
class AdaLNZero(nn.Module):
    def __init__(self, dim: int, cond_dim: int): ...
    def forward(self, x: Tensor, cond: Tensor) -> tuple[Tensor, Tensor, Tensor, Tensor]: ...
```

### Return
`forward` returns `(x_msa, gate_msa, x_mlp, gate_mlp)`:
- `x_msa = norm(x) * (1 + scale_msa) + shift_msa`
- `x_mlp = norm(x) * (1 + scale_mlp) + shift_mlp`
- gates should have shape `(B, 1, D)`

### Requirements
- Use `nn.LayerNorm(dim, elementwise_affine=False)`
- Use one linear modulation layer from `cond_dim` to `6 * dim`
- Zero-initialize the modulation layer weights and bias
