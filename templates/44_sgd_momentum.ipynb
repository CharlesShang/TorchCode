# 🟡 Medium: SGD with Momentum

Implement **stochastic gradient descent with momentum** from scratch.

For each parameter:

```python
grad = p.grad
if weight_decay > 0:
    grad = grad + weight_decay * p
velocity = momentum * velocity + grad
p = p - lr * velocity
```
