Implement simplified **MMDiT joint attention**.

MMDiT-style blocks use separate modality-specific projections for image and text tokens, then perform attention over the joint sequence.

### Signature
```python
class MMDiTJointAttention(nn.Module):
    def __init__(self, dim: int, num_heads: int): ...
    def forward(self, image_tokens: Tensor, text_tokens: Tensor) -> tuple[Tensor, Tensor]: ...
```

### Requirements
- `image_tokens`: `(B, S_img, D)`
- `text_tokens`: `(B, S_txt, D)`
- Use separate `qkv_img`, `qkv_txt`, `proj_img`, and `proj_txt`
- Concatenate projected Q/K/V across sequence length
- Return `(image_out, text_out)` with the original sequence lengths
