# 🔴 Hard: Multimodal RoPE (M-RoPE)

Implement a Qwen-style **Multimodal Rotary Position Embedding** for mixed text, image, and video tokens.

### Signature
```python
def apply_multimodal_rope(
    q: torch.Tensor,
    k: torch.Tensor,
    position_ids: torch.Tensor,
    sections: tuple[int, int, int],
    base: float = 10000.0,
) -> tuple[torch.Tensor, torch.Tensor]:
    ...
```

### Inputs
- `q`: `(B, Hq, N, D)` and `k`: `(B, Hk, N, D)`; Q and K may have different head counts
- `position_ids`: `(3, B, N)` in temporal, height, width order
- `sections`: number of adjacent rotary pairs assigned to `(t, y, x)`; it must sum to `D // 2`

### Requirements
- Treat adjacent channels `(2i, 2i + 1)` as a rotary pair
- Use `inv_freq[i] = base ** (-2i / D)` across all pairs
- Select the temporal, height, or width coordinate according to each pair's section
- Preserve shape, dtype, gradients, and per-token vector norms
- When all three position IDs are equal (text tokens), reduce exactly to ordinary 1D RoPE
- Validate tensor shapes and the section allocation
